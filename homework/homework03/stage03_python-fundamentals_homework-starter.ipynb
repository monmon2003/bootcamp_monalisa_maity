## 1. NumPy Operations

```python
# TODO: Create a NumPy array and perform vectorized operations

a = np.array([1, 2, 3, 4, 5])
print("Original array:", a)
# Vectorized operation: Squaring each element
squared_a = a ** 2
print("Squared array:", squared_a) 
# Vectorized operation: Adding 10 to each element
added_a = a + 10
print("Array after adding 10:", added_a)
```

## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit

%timeit a ** 2
def square(arr):
    for i in range(len(arr)):
        arr[i] = arr[i] ** 2
%timeit square(a.copy())
```
