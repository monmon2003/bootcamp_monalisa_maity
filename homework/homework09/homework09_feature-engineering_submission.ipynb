### Feature: `spend_income_ratio`

**Rationale:** Raw spend alone doesn't account for a person's financial capacity. $500 in monthly spend means something very different for someone earning $2,000/month versus $20,000/month. Dividing spend by income normalizes for this, producing a relative measure of spending intensity that's more comparable across individuals with different income levels. This is useful for identifying who is spending a disproportionate share of their income, which raw spend alone would obscure.

### Feature: `rolling_spend_mean`

**Rationale:** Monthly spend is noisy, a single month's value can spike due to a one-off purchase or dip due to an unusually quiet period, which doesn't necessarily reflect a person's typical spending behavior. Taking a 3-period rolling mean smooths out this short-term noise and gives a more stable estimate of underlying spending level, making it easier to spot genuine trends or shifts in behavior rather than reacting to single-month outliers.

**Note:** The first 2 rows will be `NaN` since a 3-period rolling window needs at least 3 prior observations to compute and this is expected and should be handled (e.g., dropped or filled) before using this feature downstream.
